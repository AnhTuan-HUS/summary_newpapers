# Crawl raw data fromm url

```python
import requests
import os
```

```python
url = "https://vnexpress.net/unitree-trinh-dien-robot-hinh-nguoi-tu-chien-dau-5118417.html"

response = requests.get(url)
print(response.headers)
```

```python
# lay tieu de cuar bai viet
print(response.text.split("<title>")[1].split("</title>")[0])
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')  
# lay noi dung cua bai viet
content = soup.find_all('p', class_='Normal')
print(content)
# lay link cua anh, video trong content
video_links = []
for video in soup.find_all('video'):
    video_links.append(video.get('src'))
print(video_links)
img_links = []
for img in soup.find_all('img'):
    img_links.append(img.get('src'))
print(img_links)
```

# Kiểm tra output của generics parser

```python
import sys
import json
import requests

sys.path.insert(0, "/home/mr-tuan/Projects/news_summary")

from scripts.crawler.parsers.generic_parser import GenericParser
from scripts.crawler.config_loader import load_source_config

config_path = "/home/mr-tuan/Projects/news_summary/scripts/configs/sources/rss/vnexpress.yaml"
url = "https://vnexpress.net/robot-hinh-nguoi-dau-tien-tu-buoc-ra-tu-day-chuyen-san-xuat-5118285.html"

config = load_source_config(config_path)
parser = GenericParser(config)

html = requests.get(url, timeout=20).text
article = parser.parse(html)

print(article["title"])
print(article["author"])
print(article["published_at"])
print(article["content_raw"][:200])

with open("article.json", "w", encoding="utf-8") as f:
    json.dump(article, f, ensure_ascii=False, indent=4)
```
