```python
from google import genai
from google.genai import types
from qdrant_client import QdrantClient
```

```python
GEMINI_API = os.getenv("LLM_API_KEY")
QDRANT_HOST = "localhost"
QDRANT_PORT = 6333
COLLECTIONS_NAME = "qdrant-collections"
```

```python
gemini_client = genai.Client(api_key=GEMINI_API)
qdrant_client = QdrantClient(host=QDRANT_HOST, port=QDRANT_PORT)
```

```python
promt = "Cựu lãnh đạo Google cảnh báo điều gì?"
```

```python
response = gemini_client.models.embed_content(
    model="gemini-embedding-2-preview",
    contents=promt,
    config=types.EmbedContentConfig(task_type="RETRIEVAL_QUERY")
)

# Lấy phần tử đầu tiên [0] của list 'embeddings'[cite: 1, 2]
query_vector = response.embeddings[0].values
```

```python
# truy xuaast duwx lieuj trong qdrant
results = qdrant_client.query_points(
    collection_name=COLLECTIONS_NAME,
    query=query_vector,
    limit = 3,
    with_payload=True
).points

for i in results:
    print(i.payload["page_content"])
```
