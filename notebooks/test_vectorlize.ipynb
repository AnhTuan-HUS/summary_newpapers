```python
split_text = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 100
)
```

```python
import os
from pathlib import Path
from dotenv import load_dotenv

env_path = Path.cwd().resolve().parent/ '.env'
load_dotenv(env_path)

model_name = "models/" + os.getenv("EMBEDDING_MODEL")
model = GoogleGenerativeAIEmbeddings(model= model_name, api_key=os.getenv("LLM_API_KEY"))
```

```python
# create client qdrant
client_qdrant = QdrantClient(host= "localhost", port= 6333)

#create collection
collection_name = [col.name for col in client_qdrant.get_collections().collections]

print(collection_name)

if "test" not in collection_name:
    client_qdrant.create_collection(
        collection_name="test", 
        vectors_config=models.VectorParams(
            size=3072,
            distance=models.Distance.COSINE
        )
    )
else:
    print("Collection are exist")
```

```python
print(client_qdrant.get_collections())
```

```python
# split text
from langchain_core.documents import Document

chunk = split_text.split_text(
    text
)
docs : list[Document] = []
for idx, chunk_text in enumerate(chunk):
    chunk_idx = {
        "chunk_idx": idx,
        "content": chunk_text       
    }

    doc = Document(
        page_content="hehe",
        metadata= chunk_idx
    )
    docs.append(doc)
```

```python
for i in docs:
    print(i)
```

```python
# tạo vectorstore
import uuid
from langchain_core.documents import Document


vector_store = QdrantVectorStore(
    client=client_qdrant,
    collection_name="test",
    embedding=model
)

# tạo vector id
vector_ids = [str(uuid.uuid4()) for _ in docs]
vector_store.add_documents(documents=docs, ids=vector_ids)
```
